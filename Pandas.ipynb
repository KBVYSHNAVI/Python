```python
#lets create a series using a dictionary
fruit_protein = {
    "Avacado": 2.0,
    "Guava" : 2.6,
    "Blackberries" : 2.0,
    "Orange" : 0.9,
    "Banana" : 1.1,
    "Apples" : 0.3,
    "Kiwi" : 1.1,
    "Promogranate" : 1.7,
    "Mango" : 0.8,
    "Cherrires" : 1.0
}
```

```python
s2 = pd.Series(fruit_protein)
s2
```

```python
s2 = pd.Series(fruit_protein,name = "Protein")
s2
```

```python
#Conditional Selection
s2>1
```

```python
s2[s2>1]
```

```python
#logical Operators:and,or,not
#logical and :
      # True & True->True
      #True & False ->False
      #False & False -> False
#logical or:
      #True || True -> True
       #True || False ->True
       #False || False ->False
```

```python
s2[(s2>0.5)&(s2<2)]
```

```python
s2[(s2>0.5)|(s2<=2)]
```

```python
#not operation
s2[~(s2>1)]
```

```python
#filling those null values
import pandas as pd
data = {
    "Name" :["yshu","harshitha","nithya","vasavi","priya","hasini"],
    "Age" : [10,20,30,40,50,60],
    "Department" : ["HR","IT","Cloud Engineer","Devops Engineer","Finance","HR"],
    "Salary" : [250000,400000,260000,630000,58000,69000]
}
df = pd.DataFrame(data)
df.fillna(0)
print(df)
```

```python
df["Age"].fillna(df["Age"].mean())
```

```python
df["Salary"].fillna(df["Salary"].median)
```

```python
df["Age"].fillna(method = "ffill")
```

```python
df
```

```python
df["Name"].replace("yshu","priya")
```

```python
#duplicates
df_dup = df[df.duplicated()]
df_dup
```

```python
df
```

```python
df_dup = df[df.duplicated(keep = "first")]
df_dup
```

```python
df_dup = df[df.duplicated(keep = "last")]
df_dup
```

```python
df["Age"] = df["Age"].apply(lambda x: x *3 )

print(df)
```

```python
df["Age"] = df["Age"].apply(lambda x: x /2 )

print(df)
```

```python
#joins and merges
import pandas as pd
department_info={
    "Dept" : ["HR","It","Finance"],
    "Location" : ["New York","Chicago","India"],
    "Manager" : ["Laura","Steve","Nina"]
}
df2 = pd.DataFrame(department_info)
df2
```

```python
#concatenation
pd.concat([df,df2])
```

```python
pd.concat([df,df2],axis=1)
```

```python
pd.merge(df, df2, left_on="Department", right_on="Dept")
```
